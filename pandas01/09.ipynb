```python
import pandas as pd
import matplotlib.pyplot as plt
arquivo = "../data/kc_house_data.csv"
df = pd.read_csv(arquivo, sep=",", header=0)
```

```python
df.head(2)
```

```python
##o dado foi agrupado em 50 barras, pra ver a distribuição de preços
#deu pra ver que os moveis ta na faixa de 100k a 1MM

df["price"].hist(bins=50, color="red")

plt.title("Distribuição de Preços")
plt.xlabel("Preço")
plt.ylabel("Frequência")
plt.show()
```

```python

##parametro alpha é a opacidade da cor

#maioria dos imoveis; 
df[["bedrooms", "bathrooms"]].hist(bins=30, alpha=0.4, color="Green")
```

```python
df["bedrooms"].mean()
df["bedrooms"].max()
df["bedrooms"].min()
df["bedrooms"].std()
df["bedrooms"].skew()

```
