```python
import pandas as pd
arquivo = "../data/kc_house_data.csv"
df = pd.read_csv(arquivo, sep=",", header=0)

```

```python
#total de linhas com valores ausentes
df.isnull().sum()
```

```python
#Remove todas as linhas que tem um registro faltante
#vai remover as duas linhas
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
#o how all ele só vai remover linhas que esteja dados faltantes em todas as colunas e nao só apenas em 1;

df.dropna(how="all", inplace=True)
```

```python
#atribuindo um valor média para os valores faltantes

df["floors"].fillna(df["floors"].mean(), inplace=True)
```
