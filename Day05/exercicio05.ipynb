```python
df_transacao["data"] = pd.to_datetime(df_transacao["DtCriacao"])
```

```python
first = (df_transacao.sort_values("DtCriacao")
         .drop_duplicates(keep="first", subset=["IdCliente", "data"]))

first["fonte"] = "first"
```

```python
last = (df_transacao.sort_values("DtCriacao")
         .drop_duplicates(keep="last", subset=["IdCliente", "data"]))

last["fonte"] = "last"
```

```python
df_concatenado  = pd.concat([first,last])
```
