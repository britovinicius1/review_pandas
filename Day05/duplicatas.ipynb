```python
import pandas as pd

df = pd.DataFrame({
    "nome": ["teo", "lara", "nah", "bia", "mah", "lara", "mah", "mah", 'mah'],
    "sobrenome": ["calvo", "calvo", "ataide", "ataide", "silva", "silva" , "silva", "silva", 'silva'],
    "salario" :[200,300,2500,4000,6000,12000,120000,6000, 2500]
})

df
```

```python
df
```

```python
df.drop_duplicates() ### mantem a primeira e remove a última
```

```python
df
```

```python
df.drop_duplicates(subset=["nome", "sobrenome"]) ####mantem a ultima
```

```python
df = df.sort_values("salario", ascending=False)
df = df.drop_duplicates(keep='last', subset=["nome","sobrenome"])
```

```python
df_novo = df.sort_values("salario", ascending=False)
```

```python
df_novo
```

```python
df_novo.drop_duplicates(keep='first', subset=["sobrenome","nome"])
```

```python
df_novo = (df.sort_values("salario", ascending=False)
           .drop_duplicates(keep="last", subset=["sobrenome", "nome"]))

df_novo
```
