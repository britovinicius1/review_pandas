```python
df_brinquedo = pd.DataFrame(
    {
        "nome": ["Téo", None, "Nah", "Marcio"],
        "idade": [None, None, 43, 52],
        "salario": [3453, 4324, None, 5423],
    }
)
```

```python
df_brinquedo
```

```python
df_brinquedo.dropna(how="all")
```

```python
df_brinquedo.dropna(how="all", subset=["idade"]) ###dropa o na no dataframe todas as linhas tem que ser Na, mas, o critério ele vai avaliar só a coluna idade;
```

```python
df_brinquedo.dropna(how="any", subset=["idade", "salario"]) #### pelo menos 1 vai te que ser na, qnd uma for na ele tira; Ao menos uma
df_brinquedo.dropna(how="all", subset=["idade", "salario"])  ### as duas precisa ser na pra sumir
```
