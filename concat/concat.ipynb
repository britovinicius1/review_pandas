```python
import pandas as pd

df = pd.DataFrame({
    "cliente": [1, 2, 3, 4, 5],
    "nome": ["teo", "jose", "nah", "mah", "lah"]
})

df_02 = pd.DataFrame({
    "cliente": [6, 7, 8],
    "nome": ["kozato", "laura", "dan"],
    "idade": [32, 29, 31]
})

df_03 = pd.DataFrame({
    "idade": [32,33,44,55,12]
})
```

```python
dfs = pd.concat([df,df_02], ignore_index=True)
```

```python
dfs
```

```python
pd.concat([dfs,df_03])
```

```python
pd.concat([dfs,df_03], axis=1) ####tipo um joi ele junta da esquerda pra direita
```

```python
df_03 = df_03.sort_values(by="idade",
                  ascending=False).reset_index(drop=True) #ele usa o indice pra fazer o concat
df_03
```

```python
df_final= pd.concat([df,df_03], axis=1)
```
