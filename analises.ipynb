```python
import unicodedata

def city_clean(x):
    x = x.lower().strip()
    x = unicodedata.normalize("NFKD", x)
    return x.encode("ascii", errors="ignore").decode("utf-8")
```

```python

customers["customer_city"] = customers["customer_city"].apply(city_clean)
```
