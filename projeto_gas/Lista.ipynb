```python
gas_df["DATA INICIAL"] = pd.to_datetime(gas_df["DATA INICIAL"])
gas_df["DATA FINAL"] = pd.to_datetime(gas_df["DATA FINAL"])
```

```python
### 5. Crie uma nova coluna para representar o ano e o mês(aaaa-mm)
gas_df.head(3)





```

```python
gas_df["ANO"] = gas_df["DATA FINAL"].dt.year
gas_df["ANO-MES"] = gas_df["DATA FINAL"].dt.year.astype(str) + "-" + gas_df["DATA FINAL"].dt.month.astype(str).str.zfill(2)
```

```python
gas_df.head(3)
```

```python
##6. Utilizando o value_counts(), liste todos os tipos de produtos contidos na base de dados.

gas_df["PRODUTO"].value_counts()

```

```python
filtro = gas_df["PRODUTO"] == "GASOLINA COMUM"
df_gas_comum = gas_df[filtro]
```

```python
df_gas_comum.dtypes
```

```python
df_gas_comum.head(3)
```

```python
#### 8. Qual o preço médio de revenda da gasolina em agosto de 2008?
filtro = df_gas_comum["ANO-MES"] == "2008-08"
preco_medio_revenda_agosto_2008 = df_gas_comum[filtro]["PREÇO MÉDIO REVENDA"].mean()
print(f"O preço médio de revenda da gasolia comum em agosto de 2008 foi R${preco_medio_revenda_agosto_2008:.2f}")
```

```python
###### 9. Qual o preço médio de revenda da gasolina em maio de 2014 em São Paulo?
filtro = (df_gas_comum["ANO-MES"] == "2014-05") & (df_gas_comum["ESTADO"] == "SAO PAULO")
preco_gasolina_sp_maio_2014 = df_gas_comum[filtro]["PREÇO MÉDIO REVENDA"].mean()
print(preco_gasolina_sp_maio_2014)
```

```python
#### 10. Você conseguiria descobrir em qual(quais) estado(s) a gasolina ultrapassou a barreira dos R$ 5,00? E quando isso ocorreu?

filtro = df_gas_comum["PREÇO MÉDIO REVENDA"] > 5
df_gas_maior_que_5_reais = df_gas_comum[filtro]
df_gas_maior_que_5_reais[["ESTADO", "ANO-MES", 'PREÇO MÉDIO REVENDA']].sort_values(by="ANO-MES", ascending=False)
```

```python
#top estados e meses que a gasolina ficou mais cara e maior que 5 reais;

df_gas_maior_que_5_reais[["ESTADO", "ANO-MES", 'PREÇO MÉDIO REVENDA']].sort_values(by="PREÇO MÉDIO REVENDA", ascending=False).head(10)
```
