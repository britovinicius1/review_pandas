```python
transacoes = pd.read_csv("../data/transacoes.csv", sep=";")
```

```python
transacoes.head(10)
```

```python
### pelo oque vc quer agrupar???

transacoes.groupby(by=["IdCliente"]).count()

## ta aplicando o count pra cada coluna
```

```python
(transacoes.groupby(by=["IdCliente"])["IdTransacao"].count()).sort_values(ascending=False)
```

```python
(transacoes.groupby(by=["IdCliente"], as_index=False)[["IdTransacao"]].count()).sort_values(by="IdTransacao",ascending=False)


#entender o reset_index aqui
```

```python
# Calcular mais de uma estatistica
#qtde_transacoes
# total_pontos
# pontos / transacoes

# agg mesma coisa que aggregate;

summary = (transacoes.groupby(by=["IdCliente"], as_index=False).agg({

    "IdTransacao": ["count"],
    "QtdePontos": ["sum", "mean"]
}

))



```
