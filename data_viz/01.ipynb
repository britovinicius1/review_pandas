```python
import pandas as pd

```

```python
df = pd.read_csv("canadian_immegration_data.csv")
```

```python
df
```

```python
df.info()
```

```python
df.set_index("Country", inplace=True)
```

```python
df
```

```python
anos = list(map(str, range(1980, 2014)))
```

```python
print(anos)
```

```python
df_brazil = df.loc['Brazil', anos]
```

```python
df_brazil
```

```python
df_brazil_df = df_brazil.reset_index()
df_brazil_df.columns = ['ano', 'imigrantes']
```

```python
df_brazil_df
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,4))
plt.plot(df_brazil_df['ano'], df_brazil_df['imigrantes'])
colunas_anos_agregadas_x = ['1980', '1985', '1990', '1995', '2000', '2005', '2010']
eixo_y = [500,1000,1500,2000,2500,3000]

plt.xticks(colunas_anos_agregadas_x)
plt.yticks(eixo_y)
plt.xlabel('Ano')
plt.ylabel('Número de imigrantes')
plt.title('Imigração do Brasil para o Canada')

plt.show()
```

```python
### Visualizando comparação

df_comparacao = df.loc[['Brazil', 'Argentina'], anos]


```

```python
df_comparacao
```

```python
df_comparacao = df_comparacao.T
```

```python
df_comparacao.head(5)
```

```python
plt.plot(df_comparacao['Brazil'],label = 'Brasil')
plt.plot(df_comparacao['Argentina'],label ='Argentina')
plt.title('Imigração do Brasil e Argentina para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Número de imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.legend()
plt.show()
```

```python
#Fig -> Cria uma figura em branco
# Ax -> eixos da figura

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(df_brazil_df['ano'], df_brazil_df['imigrantes'], lw=3, color='r')
#frequencia de 5 em 5
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Dados de imigração Brasil para o Canada\n1980 a 2013', fontsize=18, loc='left')
ax.set_xlabel('Ano', fontsize=14)
ax.set_ylabel('Número de imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
# plt.grid(linestyle='--')
plt.show()
```

```python
##Axs - > pq terá subplots, por isso o s
#grafico um ao lado do outro 1 linha e duas colunas, dois graficos
#figsize -> largura dps altura

fig, axs = plt.subplots(1, 2, figsize=(15,5))
#pegando o primeiro grafico o 0
axs[0].plot(df_brazil_df['ano'], df_brazil_df['imigrantes'], color='red', linestyle='--', alpha=0.5 )
axs[0].set_title('Imigração do Brasil para o canada\n1980 a 2013', fontsize=18)
axs[0].set_xlabel('Ano', fontsize=14)
axs[0].set_ylabel('Número de imigrantes', fontsize=14)
axs[]
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

#grafico 2

#interpretação grafico 2 -> dois traços valor minimo e máximo
#valor minimo +-100
#valor maximo entre 1500 e 2000
#caixinha do bloxplot
#em 25% os dados foram em torno de 400
#em 50% os dados foram +- 700
#em 75% dos anos a imigração foi em torno de1000
#os pontinhos são outliers

axs[1].boxplot(df_brazil_df['imigrantes'])
axs[1].set_title('Boxplot da imigração do Brasil para o Canadá\n1980 a 2013', fontsize='18')
axs[1].set_xlabel('Brasil')
axs[1].set_ylabel('Número de imigrantes')
axs[1].grid()

plt.show()
```

```python
fig, axs = plt.subplots(2,2, figsize=(10,6))

#espaçamento horizontal e w controla vertical

fig.subplots_adjust(hspace=0.5, wspace=0.3)

fig.suptitle('Imigração dos quatro maiores paines da América do Sul para o Canadá\n1980 a 2013')

axs[0,0].plot(df.loc['Brazil', anos])
axs[0,0].set_title('Brasil')


axs[0,1].plot(df.loc['Colombia', anos])
axs[0,1].set_title('Colombia')

axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Peru', anos])
axs[1,1].set_title('Peru')

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))

for ax in axs.flat:
    ax.set_xlabel('Ano')
    ax.set_ylabel('Número de imigrantes')

ymin = 0
ymax = 7000

for ax in axs.ravel():
    ax.set_ylim(ymin,ymax)



plt.show()
```

```python
america_sul_filtro = df['Region'] == 'South America'
df[america_sul_filtro]
```

```python
america_sul_filtro_df2 = df.query('Region == "South America"')
```

```python
america_sul_filtro_df2
```

```python
fig, ax = plt.subplots(figsize=(14,5))

cores = ['royalblue', 'orange', 'forestgreen', 'orchid',
         'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal',
         'tomato']

#index -> paises
ax.barh(america_sul_filtro_df2.index, america_sul_filtro_df2['Total'], color=cores)
ax.set_title("Imigração da América do Sul para o Canadá\n1980 a 2013", fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.show()
```

```python
america_sul_sorted = america_sul_filtro_df2.sort_values(by="Total", ascending=True)
```

```python
fig, ax = plt.subplots(figsize=(14,5))

cores = ['royalblue', 'orange', 'forestgreen', 'orchid',
         'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal',
         'tomato']

#index -> paises
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)
ax.set_title("Imigração da América do Sul para o Canadá\n1980 a 2013", fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(14,5))

cores = []

for pais in america_sul_sorted.index:
    if pais == 'Brazil':
        cores.append('red')
    else:
        cores.append('silver')

#index -> paises
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)
ax.set_title("América do Sul: Brasil foi o quarto pais com mais imigrantes\npara o Canadá\n1980 a 2013", fontsize=18, loc='left')
ax.set_ylabel('')
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)


#v+20 posiçao aonde vai ser adicionado
# i posição y do texto
#str(v) -> converte o tamanho de cada barra em texto
#ha alinhamento horizontal
#va -> alinhamento vertical

for i, v in enumerate(america_sul_sorted["Total"]):
    ax.text(v+20, i, str(v), color='black', fontsize=10, ha='left', va='center' )

ax.set_frame_on(False)
ax.get_xaxis().set_visible(False)
ax.tick_params(axis='both', which='both', length=0)


plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
ax = sns.barplot(data=top_10, y=top_10.index, x="Total", hue=top_10.index, orient='h')
ax.set_title('Países com maior imigração para o Canadá\n1980 a 2013', loc='left', fontsize=16)
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')

plt.show()
```

```python
def gerar_grafico_paleta(palette):
    fig, ax = plt.subplots(figsize=(8,4))
    ax = sns.barplot(data=top_10, y=top_10.index, x='Total', orient='h', palette=palette)
    ax.set_title("Países com maior imigração para o Canadá\n1980 a 2013", loc='left', fontsize=18)
    ax.set_ylabel("")
    plt.show()
```
