```python
###06)Quem teve mais transacao de streak
import pandas as pd
```

```python
transacoes = pd.read_csv("../data/transacoes.csv", sep=";")
clientes = pd.read_csv("../data/clientes.csv", sep=";")
produtos_transacoes = pd.read_csv("../data/transacao_produto.csv", sep=";")
produtos = pd.read_csv("../data/produtos.csv", sep = ";")
```

```python
produtos_transacoes.head(5)
```

```python
##lembrando que chave nao recebe sufixos, por isso nao tem nenhum sufixo

cliente_joined = transacoes.merge(
    produtos_transacoes,
    on=["IdTransacao"],
    how="left",
    suffixes=["_x", "_y"]
)
```

```python
cliente_joined.head(2)
```

```python
cliente_joined[["IdTransacao", "IdCliente", "IdProduto"]]
```

```python
df_full = cliente_joined.merge(
    produtos,
    right_on=["IdProduto"],
    left_on=["IdProduto"],
    how = "left"
)
```

```python
df_full.head(15)
```

```python
df_full_filtrado = df_full[df_full["descProduto"] == "Presença Streak"]
```

```python
(df_full_filtrado.groupby(by=["IdCliente"])["IdTransacao"].count().sort_values(ascending=False).head(1))
```
